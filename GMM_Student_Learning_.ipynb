```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.mixture import GaussianMixture
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# 1. Upload CSV file
uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
print(df.info())

# 2. Select numerical engagement features
excluded_keywords = [
    "id", "target", "label", "class", "outcome",
    "grade", "result"
]

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

feature_columns = [
    col for col in numeric_columns
    if not any(word in col.lower() for word in excluded_keywords)
]

if len(feature_columns) < 2:
    raise ValueError(
        "Please upload a CSV containing at least two numerical engagement features."
    )

X = df[feature_columns].copy()

print("\nSelected features:", feature_columns)

# 3. Handle missing values and standardize data
imputer = SimpleImputer(strategy="median")
X_imputed = imputer.fit_transform(X)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

# 4. Find optimal number of GMM components using BIC and AIC
max_k = min(8, max(1, len(df) // 10))
k_values = range(1, max_k + 1)

bic_scores = []
aic_scores = []

for k in k_values:
    model = GaussianMixture(
        n_components=k,
        covariance_type="full",
        n_init=5,
        random_state=42
    )
    model.fit(X_scaled)
    bic_scores.append(model.bic(X_scaled))
    aic_scores.append(model.aic(X_scaled))

plt.figure(figsize=(8, 5))
plt.plot(list(k_values), bic_scores, marker="o", label="BIC")
plt.plot(list(k_values), aic_scores, marker="s", label="AIC")
plt.xlabel("Number of Components (K)")
plt.ylabel("Information Criterion")
plt.title("Selecting Optimal GMM Components")
plt.xticks(list(k_values))
plt.legend()
plt.grid(alpha=0.3)
plt.show()

best_k = list(k_values)[np.argmin(bic_scores)]
print("Optimal number of profiles:", best_k)

# 5. Fit final Gaussian Mixture Model
gmm = GaussianMixture(
    n_components=best_k,
    covariance_type="full",
    n_init=10,
    max_iter=300,
    random_state=42
)

gmm.fit(X_scaled)

# 6. Assign profiles and calculate soft probabilities
cluster_labels = gmm.predict(X_scaled)
probabilities = gmm.predict_proba(X_scaled)

results = df.copy()
results["Learning_Profile"] = cluster_labels

for i in range(best_k):
    results[f"Probability_Profile_{i}"] = probabilities[:, i]

results["Assignment_Confidence"] = probabilities.max(axis=1)

results["Assignment_Uncertainty"] = -np.sum(
    probabilities * np.log(probabilities + 1e-10),
    axis=1
)

print("\nStudent profile assignments and probabilities:")
display(results.head(10))

# 7. Summarize profiles
profile_summary = results.groupby("Learning_Profile")[feature_columns].mean()
profile_summary["Student_Count"] = results.groupby("Learning_Profile").size()

profile_summary["Percentage"] = (
    profile_summary["Student_Count"] / len(results) * 100
)

print("\nLearning Profile Summary:")
display(profile_summary.round(2))

# 8. Visualize profile characteristics
plt.figure(figsize=(12, 6))
sns.heatmap(
    profile_summary[feature_columns].T,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Engagement Metrics Across Learning Profiles")
plt.xlabel("Learning Profile")
plt.ylabel("Engagement Feature")
plt.tight_layout()
plt.show()

# 9. Visualize clusters using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(9, 6))
scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=cluster_labels,
    cmap="viridis",
    alpha=0.7
)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Student Learning Profiles Using GMM")
plt.colorbar(scatter, label="Learning Profile")
plt.grid(alpha=0.2)
plt.show()

# 10. Visualize soft cluster probabilities
sample_size = min(30, len(results))

plt.figure(figsize=(12, 5))
plt.stackplot(
    range(sample_size),
    probabilities[:sample_size].T,
    labels=[f"Profile {i}" for i in range(best_k)],
    alpha=0.8
)
plt.xlabel("Student Index")
plt.ylabel("Membership Probability")
plt.title("Soft Cluster Probabilities")
plt.ylim(0, 1)
plt.legend(loc="upper left", bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()

# 11. Evaluate model
print("\nModel Evaluation")
print("----------------")
print("Converged:", gmm.converged_)
print("Iterations:", gmm.n_iter_)
print("BIC:", round(gmm.bic(X_scaled), 2))
print("AIC:", round(gmm.aic(X_scaled), 2))
print("Mean assignment confidence:",
      round(probabilities.max(axis=1).mean(), 4))
print("Mean assignment uncertainty:",
      round(results["Assignment_Uncertainty"].mean(), 4))

if len(np.unique(cluster_labels)) > 1:
    print(
        "Silhouette Score:",
        round(silhouette_score(X_scaled, cluster_labels), 4)
    )

```
