```python
numeric_df = df.select_dtypes(include=np.number)
```

```python
numeric_df.columns
```

```python
numeric_df.describe()
```

```python
mean = numeric_df.mean()

print("MEAN")
print(mean)
```

```python
median = numeric_df.median()

print("MEDIAN")
print(median)
```

```python
mode = numeric_df.mode().iloc[0]

print("MODE")
print(mode)
```

```python
minimum = numeric_df.min()

print("MINIMUM")
print(minimum)
```

```python
maximum = numeric_df.max()

print("MAXIMUM")
print(maximum)
```

```python
data_range = numeric_df.max() - numeric_df.min()

print("RANGE")
print(data_range)
```

```python
variance = numeric_df.var()

print("VARIANCE")
print(variance)
```

```python
std_deviation = numeric_df.std()

print("STANDARD DEVIATION")
print(std_deviation)
```

```python
quartiles = numeric_df.quantile([0.25, 0.50, 0.75])

print("QUARTILES")
print(quartiles)
```

```python
percentiles = numeric_df.quantile(
    [0.01, 0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
)

print("PERCENTILES")
print(percentiles)
```

```python
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)

IQR = Q3 - Q1

print("INTERQUARTILE RANGE (IQR)")
print(IQR)
```

```python
skewness = numeric_df.skew()

print("SKEWNESS")
print(skewness)
```

```python
kurtosis = numeric_df.kurtosis()

print("KURTOSIS")
print(kurtosis)
```

```python
cv = (numeric_df.std() / numeric_df.mean()) * 100

print("COEFFICIENT OF VARIATION (%)")
print(cv)
```

```python
statistics = pd.DataFrame({
    'Mean': numeric_df.mean(),
    'Median': numeric_df.median(),
    'Mode': numeric_df.mode().iloc[0],
    'Minimum': numeric_df.min(),
    'Maximum': numeric_df.max(),
    'Range': numeric_df.max() - numeric_df.min(),
    'Variance': numeric_df.var(),
    'Standard Deviation': numeric_df.std(),
    'Q1': numeric_df.quantile(0.25),
    'Q2': numeric_df.quantile(0.50),
    'Q3': numeric_df.quantile(0.75),
    'IQR': numeric_df.quantile(0.75) - numeric_df.quantile(0.25),
    'Skewness': numeric_df.skew(),
    'Kurtosis': numeric_df.kurtosis(),
    'Coefficient of Variation (%)':
        (numeric_df.std() / numeric_df.mean()) * 100
})

statistics
```

```python
statistics.to_csv('EV_Descriptive_Statistics.csv')

print("File saved successfully.")
```
